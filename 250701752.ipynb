```python
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose

# Load dataset
data = pd.read_csv(
    'website_traffic.csv',
    parse_dates=['Date'],
    index_col='Date'
)

# Display first few rows
print("First 5 rows:")
print(data.head())

# Time Series Decomposition
decomposition = seasonal_decompose(
    data['Page_Views'],
    model='additive',
    period=7
)

decomposition.plot()
plt.show()

# Moving Average
data['Page_Views_MA'] = data['Page_Views'].rolling(window=7).mean()

data[['Page_Views', 'Page_Views_MA']].plot(figsize=(10, 6))

plt.title('Page Views with Moving Average')
plt.xlabel('Date')
plt.ylabel('Page Views')
plt.show()

# Anomaly Detection
data['Page_Views_Z'] = (
    data['Page_Views'] - data['Page_Views'].mean()
) / data['Page_Views'].std()

# Find anomalies
anomalies = data[data['Page_Views_Z'].abs() > 3]

# Plot anomalies
plt.figure(figsize=(10, 6))

plt.plot(
    data.index,
    data['Page_Views'],
    label='Page Views'
)

plt.scatter(
    anomalies.index,
    anomalies['Page_Views'],
    color='red',
    label='Anomalies'
)

plt.title('Website Traffic and Anomalies')
plt.xlabel('Date')
plt.ylabel('Page Views')
plt.legend()
plt.grid(True)
plt.show()

# Display anomalies
print("Detected Anomalies:")
print(anomalies)
```

**Experiment 7:. Random Sampling and Sampling Distribution**

**Objective:
To explore random sampling from a population and understand the concept of sampling distribution
using Python in Jupyter Notebook.**

**Steps:**
**1. Generate a Population:
○ Create a population of data with a specified distribution (e.g., normal distribution).
2. Random Sampling:
○ Perform random sampling from the population to create multiple samples of
different sizes.
○ Compute sample statistics (mean, standard deviation, etc.) for each sample.
3. Sampling Distribution:
○ Plot histograms or density plots of sample statistics (e.g., sample means).
○ Compare the sampling distribution of the sample statistic (mean) with the
population distribution.
4. Central Limit Theorem (Optional):
○ Demonstrate the Central Limit Theorem by showing that as sample size increases,
the sampling distribution of the sample mean approaches a normal distribution
regardless of the population distribution.**


```python
import numpy as np
import matplotlib.pyplot as plt

# Set random seed for reproducibility
np.random.seed(42)

# -----------------------------------------
# 1. Generate a Population
# -----------------------------------------

# Create a normally distributed population
population = np.random.normal(
    loc=50,
    scale=10,
    size=10000
)

print("Population Mean:", population.mean())
print("Population Standard Deviation:", population.std())

# Plot population distribution
plt.figure(figsize=(10, 6))
plt.hist(population, bins=30, color='skyblue', edgecolor='black')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Population Distribution')
plt.show()


# -----------------------------------------
# 2. Random Sampling
# -----------------------------------------

sample_size = 30
number_of_samples = 1000

sample_means = []
sample_std = []

for i in range(number_of_samples):

    # Select a random sample
    sample = np.random.choice(
        population,
        size=sample_size,
        replace=False
    )

    # Calculate sample mean
    sample_means.append(np.mean(sample))

    # Calculate sample standard deviation
    sample_std.append(np.std(sample, ddof=1))

print("\nFirst 10 Sample Means:")
print(sample_means[:10])

print("\nMean of Sample Means:")
print(np.mean(sample_means))

print("\nStandard Deviation of Sample Means:")
print(np.std(sample_means))


# -----------------------------------------
# 3. Sampling Distribution
# -----------------------------------------

plt.figure(figsize=(10, 6))

plt.hist(
    sample_means,
    bins=30,
    color='lightgreen',
    edgecolor='black'
)

plt.axvline(
    np.mean(population),
    color='red',
    linestyle='--',
    label='Population Mean'
)

plt.xlabel('Sample Mean')
plt.ylabel('Frequency')
plt.title('Sampling Distribution of Sample Means')
plt.legend()
plt.show()


# -----------------------------------------
# 4. Central Limit Theorem
# -----------------------------------------

sample_sizes = [5, 30, 100]

plt.figure(figsize=(15, 4))

for j, size in enumerate(sample_sizes):

    means = []

    for i in range(1000):

        sample = np.random.choice(
            population,
            size=size,
            replace=False
        )

        means.append(np.mean(sample))

    plt.subplot(1, 3, j + 1)

    plt.hist(
        means,
        bins=30,
        color='orange',
        edgecolor='black'
    )

    plt.title('Sample Size = ' + str(size))
    plt.xlabel('Sample Mean')
    plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

Experiment 8: Conduct Z test for the Data Given using Python Objective:
To test whether the average weight of a species of birds differs from 150 grams.
Procedure:
1. Null Hypothesis (H0): The average weight of the birds is 150 grams.
2. Alternative Hypothesis (H1): The average weight of the birds is not 150 grams.
3. Sample: Measure the weights of 30 birds randomly selected from the population.
4. Z-Test: Conduct a Z-test to compare the sample mean to 150 grams.
5. Decision Rule: Use a significance level of α = 0.05.

```python
import numpy as np
import scipy.stats as stats

# Define the sample data (hypothetical weights in grams)
sample_data = np.array([
    152, 148, 151, 149, 147, 153, 150, 148, 152, 149,
    151, 150, 149, 152, 151, 148, 150, 152, 149, 150,
    148, 153, 151, 150, 149, 152, 148, 151, 150, 153
])

# Population mean under the null hypothesis
population_mean = 150

# Calculate sample statistics
sample_mean = np.mean(sample_data)
sample_std = np.std(sample_data, ddof=1)

# Number of observations
n = len(sample_data)

# Calculate the Z-statistic
z_statistic = (sample_mean - population_mean) / (sample_std / np.sqrt(n))

# Calculate the p-value
p_value = 2 * (1 - stats.norm.cdf(np.abs(z_statistic)))

# Print results
print(f"Sample Mean: {sample_mean:.2f}")
print(f"Z-Statistic: {z_statistic:.4f}")
print(f"P-Value: {p_value:.4f}")

# Decision based on significance level
alpha = 0.05

if p_value < alpha:
    print("Reject the null hypothesis: The average weight is significantly different from 150 grams.")
else:
    print("Fail to reject the null hypothesis: There is no significant difference in average weight from 150 grams.")
```

**Exp No:12 Experiment to detect outliers in a given data set.**

**Description: Understand the procedure to identify the outliers in a given dataset**

```python
# Outlier Detection using Percentile and IQR

import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Generate 16 random numbers between 1 and 100
array = np.random.randint(1, 100, 16)

print("Original Array:")
print(array)

# Calculate mean
print("\nMean:")
print(array.mean())

# Calculate percentiles
print("\n25th Percentile (Q1):")
print(np.percentile(array, 25))

print("\n50th Percentile (Median):")
print(np.percentile(array, 50))

print("\n75th Percentile (Q3):")
print(np.percentile(array, 75))

print("\n100th Percentile:")
print(np.percentile(array, 100))


# Function for outlier detection
def outDetection(array):
    Q1, Q3 = np.percentile(array, [25, 75])

    IQR = Q3 - Q1

    # Lower Range
    lr = Q1 - (1.5 * IQR)

    # Upper Range
    ur = Q3 + (1.5 * IQR)

    return lr, ur


# Find lower and upper range
lr, ur = outDetection(array)

print("\nLower Range:")
print(lr)

print("\nUpper Range:")
print(ur)


# Plot original data
plt.figure(figsize=(8, 5))
sns.histplot(array, kde=True)
plt.title("Distribution of Original Data")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()


# Remove outliers
new_array = array[(array > lr) & (array < ur)]

print("\nArray after removing outliers:")
print(new_array)


# Plot data after removing outliers
plt.figure(figsize=(8, 5))
sns.histplot(new_array, kde=True)
plt.title("Distribution After Removing Outliers")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()


# Calculate new lower and upper range
lr1, ur1 = outDetection(new_array)

print("\nNew Lower Range:")
print(lr1)

print("\nNew Upper Range:")
print(ur1)


# Final array
final_array = new_array[
    (new_array > lr1) & (new_array < ur1)
]

print("\nFinal Array:")
print(final_array)


# Final distribution
plt.figure(figsize=(8, 5))
sns.histplot(final_array, kde=True)
plt.title("Final Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

Exp No:14 Experiment to understand feature scaling.<bold>

Description: Understand the importance of feature scaling<bold>

```python
import numpy as np
import pandas as pd

# ==========================================
# EXPERIMENT NO. 14 - FEATURE SCALING
# ==========================================

# Create the dataset
data = {
    'Country': ['France', 'Spain', 'Germany', 'Spain', 'Germany',
                'France', 'Spain', 'France', np.nan, 'France'],
    'Age': [44, 27, 30, 38, 40, 35, np.nan, 48, 50, 37],
    'Salary': [72000, 48000, 54000, 61000, np.nan, 58000,
               52000, 79000, 83000, 67000],
    'Purchased': ['No', 'Yes', 'No', 'No', 'Yes',
                  'Yes', 'No', 'Yes', 'No', 'Yes']
}

# Create DataFrame
df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# ==========================================
# STEP 1: HANDLE MISSING COUNTRY
# ==========================================

df['Country'] = df['Country'].fillna(df['Country'].mode()[0])

# Separate features and target
features = df.iloc[:, :-1].values
label = df.iloc[:, -1].values

# ==========================================
# STEP 2: HANDLE MISSING AGE AND SALARY
# ==========================================

from sklearn.impute import SimpleImputer

# Age imputation
age_imputer = SimpleImputer(strategy='mean')
features[:, [1]] = age_imputer.fit_transform(features[:, [1]])

# Salary imputation
salary_imputer = SimpleImputer(strategy='mean')
features[:, [2]] = salary_imputer.fit_transform(features[:, [2]])

print("\nAfter Imputation:")
print(features)

# ==========================================
# STEP 3: ONE-HOT ENCODING OF COUNTRY
# ==========================================

from sklearn.preprocessing import OneHotEncoder

oh = OneHotEncoder(sparse_output=False)

Country = oh.fit_transform(features[:, [0]])

print("\nOne-Hot Encoded Country:")
print(Country)

# ==========================================
# STEP 4: CREATE FINAL FEATURE SET
# ==========================================

final_set = np.concatenate(
    (Country, features[:, [1, 2]]),
    axis=1
)

print("\nFinal Feature Set:")
print(final_set)

# ==========================================
# STEP 5: STANDARDIZATION
# ==========================================

from sklearn.preprocessing import StandardScaler

sc = StandardScaler()

feat_standard_scaler = sc.fit_transform(final_set)

print("\nStandard Scaled Data:")
print(feat_standard_scaler)

# ==========================================
# STEP 6: MIN-MAX SCALING
# ==========================================

from sklearn.preprocessing import MinMaxScaler

mms = MinMaxScaler(feature_range=(0, 1))

feat_minmax_scaler = mms.fit_transform(final_set)

print("\nMin-Max Scaled Data:")
print(feat_minmax_scaler)

# ==========================================
# RESULT
# ==========================================

print("\n==========================================")
print("Feature scaling completed successfully.")
print("==========================================")
```

```python
import numpy as np
import pandas as pd

# Create sample dataset
data = {
    'Country': ['France', 'Spain', 'Germany', 'Spain', 'Germany',
                'France', 'Spain', 'France', 'France', np.nan],
    'Age': [44, 27, 30, np.nan, 40, 35, 38, 48, 50, 37],
    'Salary': [72000, 48000, 54000, 61000, np.nan,
               58000, 52000, 79000, 83000, 67000],
    'Purchased': ['No', 'Yes', 'No', 'No', 'Yes',
                  'Yes', 'No', 'Yes', 'No', 'Yes']
}

df = pd.DataFrame(data)

# Display dataset information
print("Original Dataset:")
print(df)
print("\nDataset Information:")
df.info()

# Handle missing values
df['Country'] = df['Country'].fillna(df['Country'].mode()[0])
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Salary'] = df['Salary'].fillna(round(df['Salary'].mean()))

# Encode Country
updated_dataset = pd.concat(
    [pd.get_dummies(df['Country']), df[['Age', 'Salary', 'Purchased']]],
    axis=1
)

# Encode Purchased
updated_dataset['Purchased'] = updated_dataset['Purchased'].map(
    {'No': 0, 'Yes': 1}
)

# Display final dataset
print("\nPreprocessed Dataset:")
print(updated_dataset)
```

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix, classification_report

# Load Iris dataset without uploading a file
iris = load_iris()

df = pd.DataFrame(iris.data, columns=[
    'sepal.length', 'sepal.width', 'petal.length', 'petal.width'
])
df['variety'] = iris.target_names[iris.target]

# Display dataset information
df.info()

# Display class counts
print(df['variety'].value_counts())

# Display first five rows
print(df.head())

# Separate features and labels
features = df.iloc[:, :-1].values
label = df.iloc[:, -1].values

# Split dataset
xtrain, xtest, ytrain, ytest = train_test_split(
    features, label, test_size=0.33, random_state=42
)

# Create and train KNN model
model_KNN = KNeighborsClassifier(n_neighbors=5)
model_KNN.fit(xtrain, ytrain)

# Display training and testing accuracy
print("Training Accuracy:", model_KNN.score(xtrain, ytrain))
print("Testing Accuracy:", model_KNN.score(xtest, ytest))

# Confusion matrix for test data
predictions = model_KNN.predict(xtest)
print("Confusion Matrix:")
print(confusion_matrix(ytest, predictions))

# Classification report for test data
print("Classification Report:")
print(classification_report(ytest, predictions))
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, classification_report

# Create sample Social Network Ads dataset
np.random.seed(42)
n = 400

df = pd.DataFrame({
    'User ID': range(15624510, 15624510 + n),
    'Gender': np.random.choice(['Male', 'Female'], n),
    'Age': np.random.randint(18, 61, n),
    'EstimatedSalary': np.random.randint(15000, 150001, n)
})

# Generate Purchased values for demonstration
df['Purchased'] = (
    (df['Age'] > 40) |
    ((df['Age'] > 30) & (df['EstimatedSalary'] > 70000))
).astype(int)

print(df.head())
print(df.info())

# Encode Gender
df['Gender'] = df['Gender'].map({'Male': 1, 'Female': 0})

# Select features and target
features = df[['Gender', 'Age', 'EstimatedSalary']].values
label = df['Purchased'].values

# Split dataset
xtrain, xtest, ytrain, ytest = train_test_split(
    features, label, test_size=0.25, random_state=0
)

# Feature scaling
sc = StandardScaler()
xtrain = sc.fit_transform(xtrain)
xtest = sc.transform(xtest)

# Train Logistic Regression model
model_LR = LogisticRegression()
model_LR.fit(xtrain, ytrain)

# Predict
predictions = model_LR.predict(xtest)

# Accuracy
print("Accuracy:", model_LR.score(xtest, ytest))

# Confusion Matrix
print("Confusion Matrix:")
print(confusion_matrix(ytest, predictions))

# Classification Report
print("Classification Report:")
print(classification_report(ytest, predictions, zero_division=0))
```

Exp No:20Experiment to understand K-means clustering algorithm for
a given data set.
Description: Understand the K-means clustering algorithm for the dataset given.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans

# Load dataset directly from the internet
url = "https://raw.githubusercontent.com/hannazan/mall-customer-segmentation/main/data/Mall_Customers.csv"
df = pd.read_csv(url)

# Dataset information
df.info()
print(df.head())

# Pairplot
sns.pairplot(df)
plt.show()

# Select features
features = df.iloc[:, [3, 4]].values

# K-Means clustering
model = KMeans(n_clusters=5, random_state=42, n_init=10)
model.fit(features)

# Add cluster labels
Final = df.iloc[:, [3, 4]].copy()
Final['label'] = model.labels_
print(Final.head())

# Plot clusters
sns.FacetGrid(Final, hue="label", height=6) \
    .map(plt.scatter, "Annual Income (k$)", "Spending Score (1-100)") \
    .add_legend()
plt.show()

# Elbow method
wcss = []

for i in range(1, 11):
    model = KMeans(n_clusters=i, random_state=42, n_init=10)
    model.fit(features)
    wcss.append(model.inertia_)

plt.plot(range(1, 11), wcss, marker='o')
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
plt.title("Elbow Method")
plt.show()
```
